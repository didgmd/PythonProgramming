## `list.split()`

```python
sentence_list = ["Python is a cool programming language", "I love to program", "Programming is so cool"]
word_list = []
word_len_list = []
for sentence in sentence_list:
    for word in sentence.split(" "):
        word_list.append(word)
        word_len_list.append(len(word))

print(word_list)
print(word_len_list)
```

```python
sentence_list = ["Python is a cool programming language", "I love to program", "Programming is so cool"]
word_list = []
word_len_list = []
less_than_four = []
for sentence in sentence_list:
    for word in sentence.split(" "):
        word_list.append(word)
        word_len_list.append(len(word))
        if len(word) < 5:
            less_than_four.append(word)

print(word_list)
print(word_len_list)
print(less_than_four)
```

## `list.append()`

```python
car_list = ["benz", "bmw", "audi"]
print(f"original car list: {car_list}")
car_list.append("byd")
print(f"updated car list: {car_list}")
```

## `list.insert()`

```python
car_list.insert(2, "tesla")
print(f"car list after insertion at index 2: {car_list}")
```

## `list.remove()`

```python
car_list.remove("audi")
print(f"car list after removal: {car_list}")
```

## `list.pop()`

```python
car = car_list.pop()
print(f"car list after pop: {car_list}")
print(f"the car has been popped: {car}")
```

## 2D list

```python
two_dim_list = [[1,2,3],[4,5,6],[7,8,9]]
sum = 0
for first_dim in two_dim_list:
    print(first_dim)
    for second_dim in first_dim:
        print(second_dim)
        sum += second_dim

print(f"sum: {sum}")
```

## 可变 (Mutable) 与不可变 (Immutable)
- Mutable: list (列表), set (集合), dictionary (字典)
- Immutable: tuple (元组)

```python
car_set = {"audi", "benz", "bmw"}
print("audi" in car_set)  # 存在
print("tesla" in car_set)  # 不存在
```

## `set.add()`

```python
car_set.add("tesla")
print(f"car set after addition: {car_set}")  # 注意输出的次序:列表索引(index)固定,因此可以引用;集合则不能通过索引引用
print(car_set[1])  # 报错: TypeError: 'set' object is not subscriptable
```

## `set.remove()`

```python
car_set.remove("bmw")
print(f"car set after removal: {car_set}")
```

## 字典 (Dictionary)
- 由键 (Key) 和值 (Value) 构成
- 可通过 Key 引用 Value

```python
student_dict = {
    10080001: "Ann",
    10080002: "Bob",
    10080003: "Charlie",
    10080004: "David",
}

print(student_dict)
student_num = 10080001
print(f"the student with id {student_num} is named {student_dict[student_num]}")
```

## Adding item to dictionary

```python
student_dict[10080005] = "Emma"
student_dict[10080006] = "Francis"
print(student_dict)
```

```python
for k in student_dict.keys():
    print(f"the student with id {k} is named {student_dict[k]}")
```
