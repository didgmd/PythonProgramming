## Function (函数)
### 内置函数 (Inbuilt Functions)
  - `print()`
  - `len()`
  - `type()`
  - `int()`, `float()`, `str()`
  - `sort()`

```python
# 内置函数
list1 = [3, 1, 4, 1, 5, 9, 2]  # 未排序的列表
print("Original list:", list1)  # 显示原始列表
list1.sort()  # 对列表进行排序
print("Sorted list:", list1)  # 显示排序后的列表
list1.sort(reverse=True)  # 反向排序列表
print("Reverse sorted list:", list1)  # 显示反向排序后的列表
```

### 自定义函数 (User-defined Functions)
  - 定义函数：`def function_name(parameters):`
  - 调用函数：`function_name(arguments)`
  - 返回值：`return value`

**注意**: 函数需要先定义、后调用。即先写 `def` 语句，再调用函数。

```python
# 自定义函数
def add(a, b):  # 定义一个函数，计算两个数的和
    return a + b  # 返回两个数的和

result = add(5, 7)  # 调用函数，传入参数5和7
print("Sum:", result)  # 输出结果
```

```python
def maximum(a, b):  # 定义一个函数，返回两个数中的较大值
    if a > b:  # 如果a大于b
        return a  # 返回a
    else:  # 否则
        return b  # 返回b

max_value = maximum(10, 20)  # 调用函数，传入参数10和20
print("Maximum value:", max_value)  # 输出结果
```

```python
def print_hello(name):  # 定义一个函数，打印问候语
    print("Hello,", name)  # 打印问候语

print_hello("Alice")  # 调用函数，传入参数"Alice"
print_hello("Bob")  # 调用函数，传入参数"Bob"
```

### 形参 (Parameters) 与 实参 (Arguments)
  - 形参：定义函数时括号内的变量
  - 实参：调用函数时传入的具体值

```python
def list_total(list_numbers):  # 定义一个函数，计算列表中所有数字的总和
    total = 0  # 初始化总和为0
    for number in list_numbers:  # 遍历列表中的每个数字
        total += number  # 将数字加到总和中
    return total  # 返回总和

def list_length(list_items):  # 定义一个函数，计算列表的长度
    count = 0  # 初始化计数器为0
    for item in list_items:  # 遍历列表中的每个元素
        count += 1  # 计数器加1
    return count  # 返回计数器的值

def list_average(list_numbers):  # 定义一个函数，计算列表中数字的平均值
    total = list_total(list_numbers)  # 调用之前定义的函数，计算总和
    length = list_length(list_numbers)  # 调用之前定义的函数，计算长度
    if length == 0:  # 防止除以零错误
        return 0  # 如果列表为空，返回0
    return total / length  # 返回平均值

numbers = [1, 2, 3, 4, 5]  # 定义一个包含数字的列表

total_sum = list_total(numbers)  # 调用函数，传入数字列表
print("Total sum:", total_sum)  # 输出结果

length = list_length(numbers)  # 调用函数，传入数字列表
print("List length:", length)  # 输出结果

average = list_average(numbers)  # 调用函数，传入数字列表
print("List average:", average)  # 输出结果
```

## 全局变量与局部变量 (Global and Local Variables)
  - 全局变量：在函数外定义的变量，可以在函数内访问
  - 局部变量：在函数内定义的变量，只能在函数内访问

```python
# 全局变量与局部变量
global_var = "I am a global variable"  # 定义一个全局变量
def variable_scope():  # 定义一个函数，演示变量作用域
    local_var = "I am a local variable"  # 定义一个局部变量
    print(local_var)  # 打印局部变量
    print(global_var)  # 访问并打印全局变量

variable_scope()  # 调用函数

print(global_var)  # 打印全局变量
# print(local_var)  # 尝试打印局部变量（会导致错误，因为局部变量在函数外不可访问）
```

```python
def calculate_price(price, discount=0, tax=20):  # 定义一个函数，计算最终价格，包含默认参数
    discounted_price = price - (price * discount / 100)  # 计算折扣后的价格
    final_price = discounted_price + (discounted_price * tax / 100)  # 计算含税的最终价格
    return final_price  # 返回最终价格

final_price1 = calculate_price(100)  # 调用函数，使用默认折扣和税率
print("Final price (no discount):", final_price1)  # 输出结果

final_price2 = calculate_price(100, discount=10)  # 调用函数，使用10%的折扣和默认税率
print("Final price (10% discount):", final_price2)  # 输出结果

final_price3 = calculate_price(100, discount=10, tax=15)  # 调用函数，使用10%的折扣和15%的税率
print("Final price (10% discount, 15% tax):", final_price3)  # 输出结果
```

## 作用域 (Scope)
  - 全局作用域：全局变量的作用范围
  - 局部作用域：局部变量的作用范围

```python
def func_one(x):  # 定义第一个函数
    y = 1  # 定义局部变量y
    x = x + y  # 修改参数x
    print(f"x = {x} in func_one scope")  # 打印函数内的x值
    return x  # 返回修改后的x值

x = 3  # 定义全局变量x
y = 3  # 定义全局变量y
z = func_one(x)  # 调用函数，传入全局变量x
print(f"x = {x} in global scope")  # 打印全局变量x
print(f"y = {y} in global scope")  # 打印全局变量y
print(f"z = {z} in global scope")  # 打印全局变量z
```

```python
# 默认参数
def greet(name, greeting="Hello"):  # 定义一个函数，包含默认参数
    print(f"{greeting}, {name}!")  # 打印问候语

greet("Alice")  # 调用函数，使用默认问候语
greet("Bob", greeting="Hi")  # 调用函数，使用自定义问候语
greet("Charlie", "Good morning")  # 调用函数，使用另一个自定义问候语
```
