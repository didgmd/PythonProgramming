## type()函数: 用来检查变量的数据类型

```python
print(type(name))  # 输出: <class 'str'>
print(type(age))   # 输出: <class 'str'>
print(type(10))    # 输出: <class 'int'>
print(type(10.5))  # 输出: <class 'float'>
print(type(True))  # 输出: <class 'bool'>
```

## 通过input()获取的输入, 数据类型都是字符串(str)
- 输入25, type()显示为str, 因为实际获取的是字符串"25", 而不是整数25
- 如果需要整数, 需要使用int()进行类型转换: age = int(input("Enter your age: "))

```python
age = input("Enter your age: ") # 输入25, type()显示为str, 因为实际获取的是字符串"25", 而不是整数25
print(type(age))  # 输出: <class 'str'>
```

```python
age = int(input("Enter your age: "))  # 将输入转换为整数
print(type(age))  # 输出: <class 'int'>
```

## 类型转换(Type Casting)
- int(): 将值转换为整数类型
- float(): 将值转换为浮点数类型
- str(): 将值转换为字符串类型
- bool(): 将值转换为布尔类型, 注意: `非零数值`和`非空字符串`转换为`True`, `0`和`空字符串`转换为`False`

```python
print(bool(0))        # 输出: False
print(bool(10))       # 输出: True
print(bool(""))       # 输出: False
print(bool("Hello"))  # 输出: True
```

## 算术运算符
- 加法 `+` : 例如: 5 + 3
- 减法 `-` : 例如: 5 - 3
- 乘法 `*` : 例如: 5 * 3
- 除法 `/` : 例如: 5 / 3 (结果总是浮点数)
- 取余 `%` : 例如: 5 % 3 (返回除法的余数)
- 整除 `//` : 例如: 5 // 3 (返回除法的整数部分)
- 幂运算 `**` : 例如: 2 ** 3 (2的3次方)

## 运算符优先级
- 括号 `()` : 优先级最高, 用于改变默认的运算顺序
- 幂运算 `**` : 例如: 2**3 (2的3次方)
- 乘法 `*`, 除法 `/`, 取余 `%`, 整除 `//` : 这些运算符的优先级相同, 从左到右计算
- 加法 `+`, 减法 `-` : 这些运算符的优先级相同, 从左到右计算

## 关系运算符
- 等于 `==` : 例如: 5 == 5 (True), 5 == 3 (False)
- 不等于 `!=` : 例如: 5 != 3 (True), 5 != 5 (False)
- 大于 `>` : 例如: 5 > 3 (True), 3 > 5 (False)
- 小于 `<` : 例如: 3 < 5 (True), 5 < 3 (False)
- 大于等于 `>=` : 例如: 5 >= 5 (True), 3 >= 5 (False)
- 小于等于 `<=` : 例如: 3 <= 5 (True), 5 <= 3 (False)

## 逻辑运算符
- 与 `and` : 例如: True and False (False), True and True (True)
- 或 `or` : 例如: True or False (True), False or False (False)
- 非 `not` : 例如: not True (False), not False (True)

## `datetime.datetime(year,month,day,hour,minute,second)`: 创建一个表示特定日期和时间的datetime对象

```python
import datetime
birth_time = datetime.datetime(2000, 5, 15, 10, 30, 0) # 创建一个表示2000年5月15日10点30分的datetime对象
current_time = datetime.datetime.now() # 获取当前时间
diff = current_time - birth_time # 计算时间差
print(f"You have been alive for {diff.days} days.") # 输出你已经活了多少天
print(f"You have been alive for {diff.total_seconds() // 3600} hours.") # 输出你已经活了多少小时
```

## 大小写转换
- `str.upper()`: 将字符串转换为大写
- `str.lower()`: 将字符串转换为小写
- `str.capitalize()`: 将字符串的首字母大写, 其余字母小写

```python
str1 = "Hello"
print(str1.upper()) # 输出: "HELLO"
print(str1.lower()) # 输出: "hello"
print("heLLo".capitalize()) # 输出: "Hello"
```

## 字符串索引和切片(string indexing and slicing)
- 索引: 使用方括号`[]`和索引值访问字符串中的单个字符, 索引从0开始, 负数索引表示从字符串末尾开始计数
- 切片: 使用方括号`[]`和起始索引与结束索引的范围访问字符串的子串, 语法为`string[start:end]`, 包括起始索引但不包括结束索引

```python
str1 = "Hello"
print(str1[0])    # 输出: 'H'
print(str1[-1])   # 输出: 'o'
print(str1[1:4])  # 输出: 'ell'
print(str1[:3])   # 输出: 'Hel'
print(str1[2:])   # 输出: 'llo'
```

## len()函数: 用于获取字符串的长度(字符数)

```python
print(len("Hello")) # 输出: 5
print(str1[1:len(str1)]) # 输出: 'ello'
```
