## 遍历 (Traversal)
- 对列表、字典等数据结构中的每个元素进行访问和处理的过程。
- 常见的遍历方法包括使用循环（如 `for` 循环）。
- 遍历可以用于搜索、过滤、修改或计算数据。

```python
# 遍历 (Traversal)
list_data = [1, 2, 3, 4, 5]
for item in list_data:
    print(item)  # 输出列表中的每个元素

sum_data = 0
for item in list_data:
    sum_data += item
print("Sum:", sum_data)  # 输出列表元素的和
```

## 递归 (Recursion)
- 一种函数调用自身以解决问题的方法。
- 递归函数通常用于解决具有重复结构的问题，如树形结构或分治算法。
- 递归需要定义基准情况（终止条件）以防止无限循环。

```python
# 简单递归
def bradley_loves(count):
    if count <= 0:  # 终止条件
        return
    print("Bradley loves recursion!")  # 输出信息
    bradley_loves(count - 1)  # 递归调用，减少计数

bradley_loves(5)  # 调用递归函数5次
```

```python
# 递归求阶乘
def factorial(n):
    if n == 0 or n == 1:  # 基准情况
        return 1
    else:
        return n * factorial(n - 1)  # 递归调用

print("Factorial of 5:", factorial(5))  # 输出5的阶乘
```

```python
# 递归求斐波那契数列的第n个数
def fib(m):
    if m < 2:
        return 1
    return fib(m-1) + fib(m-2)

n = 6
print(f"fib({n}) = {fib(n)}")
```

```python
# 递归求斐波那契数列的前n个数
def fib(m):
    if m < 2:
        return 1
    return fib(m-1) + fib(m-2)

print([fib(n) for n in range(10)])
```
