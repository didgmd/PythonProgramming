## 类 (Class)
- 通过同一个类可以创建多个实例 (Instance)
- 这些实例具有相同的属性 (Attribute) 和方法 (Method)

```python
class Pet:
    def __init__(self, name, age):
        self.name = name  # 属性
        self.age = age

    def talk(self):  # 方法
        print(f"Hi, I am an instance of Pet, my name is {self.name}")

if __name__ == "__main__":
    pet_one = Pet("Dog", 5)  # 实例
    pet_two = Pet("Cat", 6)  # 实例

    pet_one.talk()
    pet_two.talk()

    print(f"Age of pet_one is {pet_one.age}")
    print(f"Age of pet_two is {pet_two.age}")
```

## 随机 (Random)
- `random.randint(min,max)` 随机选择一个 `min` 和 `max` 之间的整数
- `random.choice(list)` 从 `list` 中随机选择一个元素

```python
import random
index = random.randint(0, 2)
print(index)
mylist = ["apple", "banana", "cherry"]
print(mylist[index])
print(random.choice(mylist))
```

## 文件读写
- `file_handler = open(filename, method)`
- `with open(filename, method) as file_handler`

```python
file = open("example.txt", "r")  # 用只读方式打开文件
lines = file.read()
print(lines)
file.close()
```

```python
with open("example.txt", "w") as file:  # 用写入方式打开文件
    file.write("This is a new line\n")
    file.close()

with open("example.txt", "r") as file:
    lines = file.read()
    print(lines)
```
