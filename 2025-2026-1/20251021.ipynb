## 比较字符串大小
- "apple" < "banana"  # True
- "apple" > "banana"  # False
- "apple" == "banana"  # False

````python
print("apple" < "banana")  # True
print("apple" > "banana")  # False
print("apple" == "banana")  # False
````

## `while` 循环
- 循环：用于在满足某个条件时重复执行一段代码，直到条件不再满足为止。
- 语法：
  ```python
  while 条件:
      # 循环体
      执行的代码
  ```
- `break` 语句：用于立即退出循环，不再执行循环体中的剩余代码。
- `continue` 语句：用于跳过当前循环的剩余代码，直接进入下一次循环的条件判断。

## `if ... else` 语句
- 条件判断：根据条件的真假来决定执行哪一段代码。
- 语法：
  ```python
  if 条件:
      # 条件为真时执行的代码
  else:
      # 条件为假时执行的代码
  ```

````python
# 结合while, if和input实现一个简单的交互式程序
while True:
    user_input = input("Please enter something (type 'exit' to quit): ")
    if user_input.lower() == 'exit':
        print("Exiting the program. Goodbye!")
        break
    else:
        print(f"You entered: {user_input}")
````

## 关系表达式
- 比较两个值的大小或相等性，结果为布尔值（True 或 False）。
- 常见的关系运算符：
  - 小于 `<`
  - 大于 `>`
  - 等于 `==`
  - 不等于 `!=`
  - 小于等于 `<=`
  - 大于等于 `>=`

````python
print(30 > 20)  # True
print(10 < 5)   # False
print(15 == 15)  # True
print(25 != 30)  # True
print(18 >= 18)  # True
print(12 <= 10)  # False
````

## 逻辑表达式
- 逻辑与 `and`：当两个条件都为真时，结果为真。
- 逻辑或 `or`：当至少有一个条件为真时，结果为真。
- 逻辑非 `not`：将条件的真假取反。

````python
print(True and False)  # False
print(True or False)   # True
print(not True)        # False
````

## 求余 `%`
- 用于计算两个数相除后的余数。

````python
# 统计奇数个数
count = 0
for i in range(1, 21):
    if i % 2 != 0:
        count += 1
print("There are", count, "odd numbers between 1 and 20.")
````

## `range()` 函数
- 用于生成一个整数序列，常用于循环中。
- 语法：
  ```python
  range(起始值, 结束值, 步长)
  ```

````python
for i in range(1, 6):
    print(i)
````

````python
for j in range(1, 20, 2):
    print(j)
````

## 遍历字符串中的字符
- 使用 `for` 循环遍历字符串中的每个字符。
- 语法：
  ```python
  for 字符 in 字符串:
      # 对每个字符执行的代码
  ```

````python
for char in "Hello":
    print(char)
````

## 列表 `list`
- 列表是一种有序的可变集合，可以存储多个元素。
- 创建列表：使用方括号 `[]` 包含元素，元素之间用逗号分隔。
  ```python
  my_list = [1, 2, 3, 4, 5]
  ```
- 访问元素：通过索引访问列表中的元素，索引从 0 开始。
  ```python
    first_element = my_list[0]  # 访问第一个元素
    second_element = my_list[1]  # 访问第二个元素
  ```

````python
# 创建列表并对列表元素进行遍历
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
````

## `Indentation Error` 缩进错误
- 发生原因：代码块的缩进不一致或不正确。
- 解决方法：确保代码块的缩进一致。
- 示例：
  ```python
  for i in range(5):
  print(i)  # 这行代码会引发 Indentation Error，因为它没有正确缩进
  ```
- 正确写法：
  ```python
    for i in range(5):
        print(i)  # 这行代码正确缩进
  ```
