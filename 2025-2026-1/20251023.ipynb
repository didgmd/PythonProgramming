```python
input_string = input("Please enter a number:\n") # 提示用户输入一个数字
num = int(input_string) # 将输入的字符串转换为整数
while num != 5: # 当输入的数字不等于5时，继续循环
    input_string = input("Please enter a number:\n") # 提示用户输入一个数字
    num = int(input_string) # 将输入的字符串转换为整数
print("You entered 5, exiting the loop.")
```

```python
keep_going = True # 循环控制标志
while keep_going:
    input_string = input("Please enter a number:\n") # 提示用户输入一个数字
    num = int(input_string) # 将输入的字符串转换为整数
    if num == 5: # 当输入的数字等于5时，中止循环
        keep_going = False # 循环控制标志
print("Program has ended!")
```

```python
# 死循环
i = 1
while i < 6:
    print(i) # 由于i一直为1，循环是否进行的条件一直为真，导致循环不会结束
```

```python
i = 1
while i < 13:
    j = 1
    while j < 13:
        print(f"{i} x {j} = {i*j}")
        j += 1
    i += 1
```

```python
# 求1到100间所有奇数的和
total = 0
i = 1
while i < 100:
    if i % 2 != 0:  # 判断i是否为奇数
        total += i  # 将奇数累加到total中
    i += 1
print("1到100间所有奇数的和为:", total)
```
