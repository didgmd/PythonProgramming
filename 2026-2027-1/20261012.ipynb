## `print()` 函数（打印输出）
- 作用：把括号里的内容显示（输出）到屏幕上
- `print` 的意思是“打印”，可以理解为：让电脑把括号里的内容“说”出来
- 括号里的文字要用英文双引号 `"` 包起来，被引号包住的文字叫**字符串 (string)**
- 注意：括号 `()` 和引号 `"` 都必须是英文半角符号，不能用中文的（）和“”

```python
print("Hello")
```

## f-string（格式化字符串）
- **变量 (variable)**：一个贴了名字的“盒子”，里面存放着数据（例如 `name` 这个盒子里存着 `"Jim"`）
- 作用：把变量的值嵌入到字符串中
- 用法：在字符串的引号前加字母 `f`，字符串里用花括号 `{}` 包住变量名
- 运行时，`{}` 会被自动替换成变量里存的实际内容

```python
name = "Jim"
age = 20
print(f"My name is {name}, age is {age}")
```

## `for` 循环基础（loop basis）
- 作用：让计算机自动**重复**做同一件事，不用一行一行重复写
- `for i in range(10):` 表示循环 10 次：`i` 会依次等于 0, 1, 2, ..., 9
- `range(10)` 从 0 开始数，数到 9 结束，共 10 个数（不包含 10 本身）
- 冒号 `:` 后面带缩进的代码是**循环体**，表示每次循环都要执行的内容

```python
for i in range(10):
    print("Hello world is cool!")
```

### 打印循环变量 `i`
- 每循环一次，`i` 自动变成下一个数；打印 `i` 就能看到循环的过程

```python
for i in range(10):
    print(i)
```

## Escape character 转义字符 `\`
- 作用：用反斜杠 `\` 加一个字母，表示一个“看不见的特殊功能”
### `\n`：换行符（newline）
- 效果相当于在键盘上按一次回车键，让后面的内容换到下一行显示

```python
print("Hello\nWorld")
```

## Indentation 缩进
- 缩进：代码行开头的空格（一般按一下 Tab 键，或敲 4 个空格）
- Python 用缩进表示代码的“归属关系”：同一层的代码，缩进必须一致
- 在不该缩进的地方加了空格，就会报错 `IndentationError`（缩进错误）

```python
print("Hello")
 print("World")  # This will give error: IndentationError
```
