## 一、赋值 `=`
- `=` 不是“等于”，而是“赋值”：把右边的值装进左边的变量
- 先计算等号右侧的值，然后将其赋值给左侧的变量

```python
r = 10
A = 3.14 * r ** 2
print(A)
```

## 二、变量自增（increment）
- `x = x + 1`：先算右边 `x + 1`，再把结果存回 `x`（x 增加 1）
- `x += 1`：等价的简洁写法，效果相同

```python
x = 5
print(x)   # 输出 5

x = x + 1  # 先算右边 x + 1，再把结果存回 x
print(x)   # 输出 6

x += 1     # 等价于 x = x + 1
print(x)   # 输出 7
```

## 三、Python 的变量命名规则
- 只能由字母、数字、下划线 `_` 组成
- 不能以数字开头
- 区分大小写 (case sensitive)，即 `span` 不同于 `Span`
- 举例：`age`、`name_1` 合法；`1name`、`my-name`（含 `-`）不合法

## 四、表达式与运算优先级（Precedence）
- `( )` > `**` > `*`, `/`, `%`, `//` > `+`, `-`
- 和数学一样：括号最优先，先乘除后加减
- 不确定顺序时，加括号最保险

## 今日小结
- 赋值 `=`：先算右边，再把结果存进左边的变量
- 变量自增：`x = x + 1` 等价于 `x += 1`
- 命名规则：字母 / 数字 / 下划线，不能以数字开头，区分大小写
- 运算优先级：`( )` > `**` > `*` `/` `%` `//` > `+` `-`
