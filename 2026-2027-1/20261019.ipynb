## 一、关系运算符与关系表达式
- 关系运算符（relational operator）：比较两个值的大小或相等关系
- `==` 等于、`!=` 不等于、`>` 大于、`<` 小于、`>=` 大于等于、`<=` 小于等于
- 注意：`==` 是“比较是否相等”（两个等号），`=` 是“赋值”（一个等号），不要混淆
- 关系表达式的结果是布尔值 bool：只有 `True`（真）或 `False`（假）两种

```python
print(5 > 3)       # 大于，输出 True
print(5 < 3)       # 小于，输出 False
print(5 == 5)      # 等于（两个等号），输出 True
print(5 != 5)      # 不等于，输出 False
print(6 >= 6)      # 大于等于，输出 True
print("a" == "b")  # 字符串也可以比较，输出 False
```

## 二、逻辑运算符与逻辑表达式
- 逻辑运算符（logical operator）：把多个 True / False 组合起来
- `and` 与：两边**都**是 True，结果才是 True
- `or` 或：只要**有一边**是 True，结果就是 True
- `not` 非：取反，True 变 False，False 变 True

```python
print(True and True)    # and 与：两边都真才为真，输出 True
print(True and False)   # 输出 False
print(True or False)    # or 或：有一边真就为真，输出 True
print(not True)         # not 非：取反，输出 False

age = 20
print(age >= 18 and age <= 60)  # 组合使用：年龄在 18 到 60 之间，输出 True
```

## 三、分支 Branching
- 分支：根据条件（True / False）决定执行哪一段代码
- `if`...`else`...：条件成立执行 `if` 下面的代码，否则执行 `else` 下面的代码
- `elif`...：else if 的缩写，用于多个条件（从上往下依次检查，遇到第一个 True 就执行）
- 嵌套 Nesting：`if` 里面可以再写 `if`（缩进一层套一层）
- 注意：条件后面要有冒号 `:`，属于分支的代码要缩进

```python
# 嵌套分支：判断 x 能否被 3 和 5 整除
x = int(input("Please enter an integer number:"))

if x % 3 == 0:
    if x % 5 == 0:
        print(f"{x} can be divided by 3 and 5")   # 能同时被 3 和 5 整除
    else:
        print(f"{x} can be divided by 3 only")    # 只能被 3 整除
else:
    if x % 5 == 0:
        print(f"{x} can be divided by 5 only")    # 只能被 5 整除
    else:
        print(f"{x} cannot be divided by 3 or 5") # 都不能整除
```

```python
# 综合例子：关系运算符 + 逻辑运算符 + 分支（if / elif / else）
age = int(input("Please enter your age: "))

if age < 0 or age > 150:
    print("Invalid age!")       # 不合法的年龄
elif age < 18:
    print("You are a minor.")   # 未成年
elif age <= 60:
    print("You are an adult.")  # 成年人
else:
    print("You are a senior.")  # 老年人
```

```python
# 判定闰年（leap year）：能被 4 整除但不能被 100 整除，或者能被 400 整除
year = int(input("Please enter a year: "))

if (year % 4 == 0 and year % 100 != 0) or year % 400 == 0:
    print(f"{year} is a leap year")      # 是闰年
else:
    print(f"{year} is not a leap year")  # 不是闰年
```

## 四、`while` 循环
- `while`：当条件为 True 时，不断重复执行循环体内的代码
- 格式：`while 条件:` + 缩进的循环体
- **必须有办法让条件最终变成 False**，否则会变成无限循环
- `continue`：跳过本轮循环剩余的代码，直接进入下一轮
- `break`：立刻跳出整个循环

```python
# 例子 1：用 while 从 1 数到 5
count = 1
while count <= 5:
    print(count)    # 打印当前的数
    count += 1      # 每次加 1，让条件最终变成 False
```

```python
# 例子 2：break 提前结束循环
count = 1
while count <= 10:
    if count == 4:
        break       # 数到 4 时立刻跳出循环
    print(count)
    count += 1      # 只会打印 1, 2, 3
```

```python
# 猜数字游戏：用户不断猜一个 0~100 之间的秘密数字
secret_number = 42
input_number = 0          # 初始值设为 0，保证和 secret_number 不同，从而进入循环

while secret_number != input_number:
    input_number = int(input("Please choose an integer number between 0 and 100: "))

    if input_number > secret_number:
        print("Your guess is too big, try again!")   # 猜大了
    elif input_number < secret_number:
        print("Your guess is too small, try again!") # 猜小了
    else:
        print("Your guess is correct!")              # 猜对了
        break                                          # 猜对后跳出循环，游戏结束
```

## 今日小结
- 关系运算符：`==` `!=` `>` `<` `>=` `<=`，结果是 True / False
- 注意：`==` 是比较（两个等号），`=` 是赋值（一个等号）
- 逻辑运算符：`and`（都真才真）、`or`（一真即真）、`not`（取反）
- 分支：`if` / `elif` / `else` 根据条件选择执行，注意冒号和缩进，也可以嵌套
- `while` 循环：条件为 True 就重复执行，注意避免无限循环
- `continue`：跳过后续代码进入下一轮；`break`：立刻跳出循环
