## 一、递归 Recursion
- 递归：**函数自己调用自己**，把一个大问题拆成“一小步 + 一个更小的同类问题”
- 写递归需要两部分：
  1. **基本情况（base case）**：最简单的情况，直接给出答案，**不再调用自己**（否则会无限递归）
  2. **递归调用**：把问题缩小一点，再调用自己
- 例：阶乘 `fact(4) = 4 * fact(3)`，`fact(3) = 3 * fact(2)`……一直问到 `fact(0)` 得到 1，再逐层返回

```python
# 例1：阶乘（factorial）——n! = n * (n-1) * ... * 1
def fact(n):
    if n < 2:             # 基本情况：0! 和 1! 都等于 1，直接返回
        return 1
    return n * fact(n - 1)  # 递归调用：n! = n * (n-1)!

for i in range(5):
    print(fact(i))        # 依次输出 1, 1, 2, 6, 24
```

```python
# 例2：倒计时——打印 n 到 1，最后打印 Go!
def countdown(n):
    if n == 0:            # 基本情况：数到 0，喊出发并结束
        print("Go!")
        return
    print(n)              # 先打印当前数字
    countdown(n - 1)      # 递归调用：问题缩小为从 n-1 开始倒计时

countdown(5)              # 依次输出 5, 4, 3, 2, 1, Go!
```

```python
# 例3：斐波那契数列（Fibonacci）——每一项等于前两项之和：1, 1, 2, 3, 5, 8, ...
def fib(n):
    if n <= 1:                  # 基本情况：第 0、1 项就是 0 和 1 本身
        return n
    return fib(n - 1) + fib(n - 2)  # 递归调用：第 n 项 = 前两项之和

for i in range(8):
    print(fib(i), end=" ")    # 依次输出 0 1 1 2 3 5 8 13
```

## 今日小结
- 递归：函数自己调用自己，把大问题拆成小问题
- 必须有**基本情况**（base case）停止递归，否则会无限调用自己
- 递归调用时问题要**不断缩小**（例如 `n - 1`），否则会永远停不下来
- 例子：阶乘 `fact(n)`、倒计时 `countdown(n)`、斐波那契数列 `fib(n)`（一项 = 前两项之和）
