## 一、控制台 Console
- 控制台：可以直接输入 Python 代码、并立即看到运行结果的地方（像和计算机“对话”）
- 看到 `>>>` 符号，表示已进入 Python 控制台，可以开始输入代码了
- 进入 Console (VS Code / PyCharm): 在终端 (Terminal) 输入 python 并回车
- 退出 Console: 输入 `exit()` 并回车

## 二、数据类型与 `type()` 函数
- `type()`：查看一个值的数据类型（它属于哪一“种类”）
- 常见数据类型：
  - 整数 `int`：例如 3, -5, 0
  - 浮点数 `float`：例如 3.14, -0.5, 2.0
  - 字符串 `str`：用引号包住的文字，例如 "Hello", "123"
  - 布尔值 `bool`：只有 True（真）和 False（假）两种

```python
print(type(3))
```

```python
print(type(10))       # 整数 int
print(type(10.5))     # 浮点数 float
print(type("Hello"))  # 字符串 str
print(type(True))     # 布尔值 bool
```

## 三、语句（Statement）与表达式（Expression）
- **Expression 表达式**：能算出一个“值”的代码，例如 `1 + 2`、`"Hello " + "World"`
- **Statement 语句**：让计算机“做一件事”的一整行代码，例如赋值语句 `str1 = ...`、打印语句 `print(...)`
- 区别：表达式有“结果值”，语句表示“做一个动作”；语句里可以包含表达式

```python
# Expression（表达式）：能算出一个“值”
1 + 2                      # 值为 3
"Hello " + "World"        # 值为 "Hello World"
3 * 4 - 5                  # 值为 7

# Statement（语句）：让计算机“做一件事”
str1 = "Hello " + "World"  # 赋值语句：把表达式的值存进变量 str1
print(str1)                # 打印语句：把 str1 的值显示到屏幕
```

## 四、变量的重新赋值（不会自动更新）
- 计算机从上往下**一行一行**执行代码
- 改变 `radius` 后，之前算好的 `circumference` **不会自动跟着变**
- 必须重新执行计算那一行，结果才会更新

```python
pi = 3.14            # 变量 pi 存放圆周率
radius = 6           # 变量 radius 存放半径
circumference = 2 * pi * radius  # 计算周长，存入变量 circumference
print(circumference) # 输出 37.68

radius = 1           # 把 radius 重新赋值为 1
print(circumference) # 仍是 37.68！周长不会自动更新

circumference = 2 * pi * radius  # 用新的半径重新计算一次
print(circumference) # 输出 6.28
```

## 五、运算符 Operator
- `+` 加、`-` 减、`*` 乘、`/` 除（注意：`/` 的结果总是浮点数 float）
- `%` 取余（modulus）：返回除法的**余数**，例如 5 % 3 得 2
- `**` 幂运算（power）：例如 2 ** 3 表示 2 的 3 次方，得 8
- `//` 整除（floor division）：只保留除法结果的**整数部分**，例如 5 // 3 得 1

```python
print(5 + 3)   # 加，输出 8
print(5 - 3)   # 减，输出 2
print(5 * 3)   # 乘，输出 15
print(5 / 3)   # 除，输出 1.6666...（结果总是 float）
print(5 % 3)   # 取余，输出 2（5 除以 3 余 2）
print(5 // 3)  # 整除，输出 1（只留整数部分）
print(2 ** 3)  # 幂，输出 8（2 的 3 次方）
```

## 六、字符串（String）
### 6.1 打印双引号：转义字符 `\"`
- 想在字符串里包含双引号，要用转义字符 `\"`
- 如果直接写 `"`，Python 会误以为字符串提前结束，报 `SyntaxError`（语法错误）

```python
print("This is a correct \"Hello World\" program.")
print("This is a wrong "Hello World" program.")  # This line will give SyntaxError
```

### 6.2 多行字符串（Multi-line strings）
- 用三个引号 `""" ... """` 包住，字符串里可以直接换行
- 换行会被原样保留，打印出来也是多行

```python
# 多行字符串：用三个双引号包住，可以跨越多行
poem = """静夜思
床前明月光，疑是地上霜。
举头望明月，低头思故乡。"""
print(poem)
```

### 6.3 字符串拼接与类型错误 `TypeError`
- 用 `+` 可以把多个字符串**拼接**成一个字符串
- `"Age: " + "28"` 不报错：`"28"` 有引号，是字符串 str，字符串 + 字符串 = 拼接
- `"Age: " + 28` 报 `TypeError`：`28` 没有引号，是整数 int，字符串和整数不能直接相加
- 解决：用 `str(28)` 把整数转成字符串，或改用 f-string / `.format()`（见第七节）

```python
"Hello" + " " + "World"  # 字符串 + 字符串：不报错，结果是拼接
"Age: " + "28"  # "28" 有引号是字符串：不报错，结果是 "Age: 28"
"Age: " + 28  # 28 是整数：报 TypeError
```

## 七、`print()` 格式化输出
- `.format()` 方法：字符串里用 `{}` 占位，按顺序填入变量
- f-string：字符串前加 `f`，`{}` 里直接写变量名（更简洁，推荐）

```python
str1 = "Tom"
str2 = "Jerry"
print("{} and {} are friends!".format(str1, str2))
print(f"{str1} and {str2} are friends!")
```

## 八、常见错误之命名错误 `NameError`
- 使用了**没有定义过**的变量名，就会报 `NameError`
- 常见原因：变量名拼写错误（例如把 `str1` 误写成 `str3`），或变量还没赋值就使用

```python
print(str3)  # This gives NameError
```

## 九、时间模块 `datetime`
- 模块（module）：别人写好的“工具包”，用 `import` 导入后就能使用
- `datetime.datetime.now()`：获取当前的日期和时间

```python
import datetime
current_time = datetime.datetime.now()  # 获取当前时间
print(f"The date and time is {current_time}")
```
