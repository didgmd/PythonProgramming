## 一、列表进阶：常用方法
- 列表是可变的（mutable），创建后还能添加、删除元素
- **添加**：
  - `列表.append(元素)`：把元素添加到列表的**末尾**
  - `列表.insert(位置, 元素)`：把元素插入到**指定位置**（索引从 0 开始），后面的元素自动后移
  - `列表.extend(另一个列表)`：把另一个列表的所有元素逐个加到末尾（相当于批量 `append()`）
- **删除**：
  - `列表.remove(值)`：删除列表中**第一个**等于该值的元素；如果值不存在会报错
  - `列表.pop()`：删除并**返回**末尾的元素；`列表.pop(位置)`：删除并返回指定位置的元素
- **复制**：
  - `列表.copy()`：复制一个**内容相同但互相独立**的新列表（修改副本不影响原列表）

```python
scores = [90, 85, 77]
print(scores)             # [90, 85, 77]

scores.append(60)         # append：加到末尾
print(scores)             # [90, 85, 77, 60]

scores.insert(1, 95)      # insert：在索引 1 的位置插入 95，后面的元素后移
print(scores)             # [90, 95, 85, 77, 60]

scores.insert(0, 100)     # 插入到最前面
print(scores)             # [100, 90, 95, 85, 77, 60]

scores.extend([70, 80])   # extend：把另一个列表的元素逐个加到末尾
print(scores)             # [100, 90, 95, 85, 77, 60, 70, 80]

scores.remove(95)         # remove：删除第一个 95
print(scores)             # [100, 90, 85, 77, 60, 70, 80]

popped = scores.pop()     # pop()：删除并返回末尾的元素
print(popped)             # 80
print(scores)             # [100, 90, 85, 77, 60, 70]

backup = scores.copy()    # copy()：复制一个独立的新列表
backup.append(55)         # 修改副本……
print(backup)             # [100, 90, 85, 77, 60, 70, 55]
print(scores)             # [100, 90, 85, 77, 60, 70]（原列表不受影响）
```

### 注意：各列表方法的返回值
- `append()` / `insert()` / `extend()` / `remove()`：返回 `None`（只负责修改原列表）
- 所以不要写 `scores = scores.append(60)`，这样会把 `scores` 变成 `None`！
- `pop()`：返回被删除的元素（可以用变量保存）

## 二、循环遍历列表
### `for` 循环遍历列表
- `for 变量 in 列表:`：依次取出列表里的**每个元素**，赋给变量并执行循环体
- 不需要写索引，也不需要 `i += 1`，列表里有几个元素就循环几次
- 想同时拿到索引，可以用 `range(len(列表))`

```python
# for 循环直接遍历列表里的每个元素
scores = [90, 85, 77, 60]

for s in scores:
    print(s)              # 依次输出 90, 85, 77, 60

# 需要索引时：配合 range(len())
for i in range(len(scores)):
    print(f"第 {i + 1} 个元素是 {scores[i]}")
```

### `while` 循环遍历列表
- 也可以用 `while` + 索引变量遍历列表，但要自己写 `i += 1`（不如 `for` 简洁，了解即可）

```python
# while 循环遍历列表：自己控制索引
scores = [90, 85, 77, 60]

i = 0
while i < len(scores):
    print(f"第 {i + 1} 个元素是 {scores[i]}")
    i += 1            # 别忘了更新索引，否则会死循环
```

## 三、循环与方法的配合（两个常见套路）
### 套路1：空列表 + `for` 循环 + `append()`
- 先创建一个**空列表**，再在循环里用 `append()` 逐个收集结果

```python
# 从空列表开始，逐个添加 0~4 的平方
squares = []                  # 空列表

for i in range(5):
    squares.append(i ** 2)    # 每轮把 i 的平方加到末尾

print(squares)                # [0, 1, 4, 9, 16]
```

### 套路2：`while` 循环 + `pop()` 逐个处理元素
- 用 `while` 循环 + `pop()`，每轮取出一个元素来处理，直到列表为空

```python
# 用 while + pop() 逐个取出并处理元素，直到列表为空
tasks = ["作业", "复习", "预习"]

while len(tasks) > 0:
    task = tasks.pop()      # 每轮从末尾取出一个元素
    print(f"处理：{task}，还剩 {len(tasks)} 项")

print(tasks)                # []（列表已被取空）
```

## 四、用 `+` 将多个列表相加（拼接）
- `列表1 + 列表2`：把两个列表拼成一个**新列表**，原来的列表都不变
- 与 `extend()` 的区别：`+` 生成**新列表**（需要赋值保存）；`extend()` 直接**修改原列表**

```python
list1 = [1, 2, 3]
list2 = [4, 5, 6]

list3 = list1 + list2     # +：拼接成一个新列表，需要赋值保存
print(list3)              # [1, 2, 3, 4, 5, 6]
print(list1)              # [1, 2, 3]（原列表没有被修改）

list1.extend(list2)       # extend：直接修改 list1 本身
print(list1)              # [1, 2, 3, 4, 5, 6]
```

## 五、对列表操作的常用函数
- `max(列表)`：最大值；`min(列表)`：最小值
- `sum(列表)`：所有元素之和；`len(列表)`：元素个数
- `sorted(列表)`：返回一个**排好序的新列表**，原列表不变

```python
scores = [90, 85, 77, 60]

print(max(scores))     # 最大值 90
print(min(scores))     # 最小值 60
print(sum(scores))     # 总和 312
print(len(scores))     # 元素个数 4

print(sorted(scores))  # 排序后的新列表 [60, 77, 85, 90]
print(scores)          # 原列表不变 [90, 85, 77, 60]
```

## 六、二维列表（2 Dimensional List）
- 二维列表：列表里面再装列表，像一张有**行**和**列**的表格，例如 `[[90, 85], [77, 60]]`
- 取元素用两个索引：`列表[行][列]`，行、列编号都**从 0 开始**
- 可以用嵌套的 `for` 循环逐行逐列遍历

```python
# 二维列表：3 个学生，每人 2 门成绩（像一张 3 行 2 列的表格）
grade_table = [[90, 85],
               [77, 60],
               [88, 92]]

print(grade_table[0])      # 第 1 行：[90, 85]
print(grade_table[0][1])   # 第 1 行第 2 列：85
print(grade_table[2][0])   # 第 3 行第 1 列：88

# 嵌套循环：逐行逐列打印
for row in grade_table:
    for score in row:
        print(score, end=" ")  # end=" "：不换行，用空格分隔
    print()                    # 每行结束后换行
```

## 七、`==` 与 `is`（`!=` 与 `is not`）
- `==`：比较两个值的**内容**是否相等；`!=`：内容是否不相等
- `is`：比较两个变量是否指向**同一个对象**（内存中的同一个位置）；`is not` 反之
- 两个内容一样的列表：`==` 是 True，但 `is` 是 False（它们是两个独立的列表）
- 从内存角度理解：变量是“贴在一块数据上的标签”
  - `b = [1, 2, 3]` 新建了一块数据，`b` 贴在上面；它和 `a` 的数据在内存的**不同位置**
  - `c = a` 没有新建数据，只是把 `c` 也贴到 `a` 那块数据上，所以 `a is c` 为 True
  - 因此：通过 `c` 修改列表，`a` 看到的内容也会跟着变！想要独立副本请用 `copy()`
- 日常比较值请用 `==`；`is` 主要用于判断 `None`（例如 `x is None`）

```python
a = [1, 2, 3]
b = [1, 2, 3]   # 内容和 a 一样，但是另一个独立的列表
c = a           # c 和 a 指向同一个列表

print(a == b)      # True：内容相等
print(a is b)      # False：不是同一个对象（a 和 b 存放在内存的两个不同位置）
print(a is c)      # True：c 就是 a 本身

print(a != b)      # False：内容相等，所以不是"不等于"
print(a is not b)  # True：确实不是同一个对象

x = None
print(x is None)   # True：is 最常见的用法——判断 None
```

- 二维列表：列表里装列表（像表格），用 `列表[行][列]` 取元素，嵌套 `for` 循环遍历
