## 一、认识 Tuple 元组
- 元组：用**圆括号 `()`** 包住的一组数据，例如 `(5, 4, 3)`、`("Jim", 20)`
- 与列表类似：有顺序、可以重复、可以混合不同类型，索引用法和列表一样（`元组[编号]`，从 0 开始）
- 与列表最大的区别：**不能修改**（immutable）——不能改元素、不能增删
- 常见用途：存放一组不该被改动的数据；函数返回多个值

```python
# 例1：元组不能修改（immutable）
z = (5, 4, 3)
print(z)      # (5, 4, 3)
print(z[2])   # 3（读取元素没问题，和列表一样）

z[2] = 0  # 报错 TypeError！元组创建后不能修改元素
```

## 二、元组解包（Unpacking）
- 解包：把元组里的元素**按顺序**分别赋给多个变量
- 写法：`(x, y) = (4, 'fred')`，左边变量的个数要和元组元素个数一致

```python
# 例2：元组解包（unpacking）——把元组里的元素按顺序分别赋给多个变量
(x, y) = (4, 'fred')  # x 得到 4，y 得到 'fred'
print(x)  # 4
print(y)  # fred
```

## 三、函数可以通过 Tuple 返回多个值
- 函数 `return` 后面写上多个值（用逗号隔开），实际上就是把它们打成一个**元组**返回
- 调用时可以用解包的方式接收：`a, b = 函数(...)`，一次拿到所有返回值

```python
# 函数返回多个值（元组），调用时解包接收
def min_max(numbers):
    return min(numbers), max(numbers)  # 返回一个元组 (最小值, 最大值)

scores = [90, 85, 77, 60]
lo, hi = min_max(scores)  # 解包：lo = 60, hi = 90
print(lo, hi)             # 60 90
```

## 四、`zip` 函数
- `zip` 函数可以把多个可迭代对象（如列表、元组）“压缩”成一个个元组，返回一个可迭代对象
- 配对规则：按**位置**一一配对——第 1 个配第 1 个，第 2 个配第 2 个……
- 如果各列表长度不一样，会在**最短的那个**结束（多余的元素被忽略）
- 用 `list(...)` 把结果转成列表才能打印查看；常配合 `for` 循环 + 解包一起用

```python
# 例1：zip 两个列表——按位置一一配成元组
list1 = [1, 2, 3]
list2 = ['a', 'b', 'c']
zipped = zip(list1, list2)
print(list(zipped))   # [(1, 'a'), (2, 'b'), (3, 'c')]

# 例2：zip 多个列表——每个元组里有多个元素
list3 = [4, 5, 6]
list4 = ['d', 'e', 'f']
zipped = zip(list1, list2, list3, list4)
print(list(zipped))   # [(1, 'a', 4, 'd'), (2, 'b', 5, 'e'), (3, 'c', 6, 'f')]

# 例3：长度不一样——在最短的列表处结束
list5 = [10, 20]
print(list(zip(list1, list5)))  # [(1, 10), (2, 20)]（3 和 'c' 被忽略）

# 例4：配合 for 循环 + 解包，逐个取出配对结果
names = ['Tom', 'Lucy', 'Jim']
scores = [90, 85, 77]
for name, score in zip(names, scores):
    print(f"{name}: {score}")  # 依次输出 Tom: 90、Lucy: 85、Jim: 77
```

## 五、元组的比较与常用方法
### 5.1 比较 Tuple：逐个元素进行比较
- 元组比较大小：从**第 0 个元素**开始，逐个对比，遇到**第一对不同**就分出结果（像查字典比单词一样）
- 前面都相同时，短的元组更小；完全相等才是 `==`

```python
# 元组比较：从左到右逐个元素比，遇到第一对不同就分出结果
print((1, 2, 3) < (1, 2, 4))   # True：前两个相同，第三个 3 < 4
print((1, 5) < (2, 0))         # True：第一个 1 < 2，后面不用再比
print(('a', 'b') > ('a', 'a')) # True：第二个 'b' > 'a'（按字母顺序）
print((1, 2) < (1, 2, 0))      # True：前面都相同，短的更小
```

### 5.2 元组的常用方法和规则
- `元组.count(值)`：数某个值出现了几次；`元组.index(值)`：找某个值第一次出现的位置
- `len()`、`in`（是否包含）、`+`（拼接成新元组）、切片 `[开始:结束]` 都和列表用法一样
- 因为不可修改，所以元组**没有** `append` / `insert` / `remove` / `pop` 这些方法

```python
# 元组常用方法
t = (1, 2, 2, 3)
print(t.count(2))   # 2 出现了 2 次
print(t.index(3))   # 3 在索引 3
print(len(t))       # 4
print(2 in t)       # True
print(t + (4, 5))   # (1, 2, 2, 3, 4, 5)：拼接成新元组，t 不变
```

## 六、Dictionary 字典
- 字典：用**花括号 `{}`** 存放一组“键 key → 值 value”的对应关系，例如 `{"Jim": 90, "Lucy": 85}`
- 规则：键不能重复、必须是不可变类型（字符串、数字、元组）；值可以是任何类型
- 访问和修改都用 `字典[键]`；用不存在的键取值会报 `KeyError`
- `in`：检查**键**是否存在；`del 字典[键]`：删除一个键值对
- 更安全的取值方法 `get()` 和更多常用操作，见下面两节

```python
scores = {"Jim": 90, "Lucy": 85}

print(scores["Jim"])        # 90：按键取值
scores["Tom"] = 77          # 键不存在：新增一个键值对
scores["Jim"] = 95          # 键已存在：修改对应的值
print(scores)               # {'Jim': 95, 'Lucy': 85, 'Tom': 77}

print("Tom" in scores)      # True：in 检查的是键

del scores["Lucy"]          # 删除键值对
print(scores)               # {'Jim': 95, 'Tom': 77}
```

## 七、字典 `get()` 详解
- `字典[键]`：键不存在时会**报错** `KeyError`；`字典.get(键)`：键不存在时返回 `None`，**不报错**
- `字典.get(键, 默认值)`：键不存在时返回你指定的默认值
- 经典用法：配合 `if` 或计数，例如统计每个元素出现的次数

```python
scores = {"Jim": 90, "Lucy": 85}

# 例1：get() 安全取值——键不存在时返回 None，不会报错
print(scores["Jim"])       # 90
print(scores.get("Amy"))   # None（如果写 scores["Amy"] 会报 KeyError）

# 例2：get(键, 默认值)——键不存在时用默认值代替
print(scores.get("Amy", 0))   # 0

# 例3：经典用法——统计每个水果出现的次数
fruits = ["apple", "banana", "apple", "orange", "banana", "apple"]
counts = {}
for f in fruits:
    counts[f] = counts.get(f, 0) + 1  # 没见过就从 0 开始，见过就在原数上加 1
print(counts)  # {'apple': 3, 'banana': 2, 'orange': 1}
```

## 八、字典的常用操作
- `字典.keys()` / `字典.values()` / `字典.items()`：所有键 / 所有值 / 所有键值对
- `字典1.update(字典2)`：把字典2的键值对合并进来，相同键会被覆盖
- `字典.pop(键)`：删除并**返回**该键对应的值（和列表的 `pop()` 类似）
- `len(字典)`：键值对的个数
- 遍历：`for k, v in 字典.items():`，用解包同时拿到键和值

```python
scores = {"Jim": 90, "Lucy": 85}

# update()：合并另一个字典，相同的键被覆盖
scores.update({"Tom": 77, "Jim": 95})
print(scores)            # {'Jim': 95, 'Lucy': 85, 'Tom': 77}

# pop(键)：删除并返回该键的值
v = scores.pop("Tom")
print(v)                 # 77
print(scores)            # {'Jim': 95, 'Lucy': 85}

print(len(scores))       # 2：键值对的个数

print(scores.keys())       # 所有键
print(scores.values())     # 所有值

# for + items() 遍历：用解包同时拿到键和值
for name, score in scores.items():
    print(f"{name}: {score}")  # 依次输出 Jim: 95、Lucy: 85
```
