## 一、Infinite loop 死循环
- 死循环：循环条件**永远为 True**，程序一直运行、永不停止
- 常见原因：循环体内没有更新条件的语句（例如下面的 `i` 永远是 0，`i < 3` 永远成立）
- 修复方法：在循环体中加上更新条件的代码，例如 `i += 1`
- 检查方法：写 `while` 前先问自己——“循环体里有没有让条件最终变成 False 的代码？”

```python
# 错误示例：死循环（演示用，如运行需手动点击“停止”按钮）
i = 0
while i < 3:
    print(i)  # i 永远不变，条件永远为 True，会一直打印 0
```

```python
# 正确示例：每轮把 j 加 1，条件最终变成 False，循环正常结束
j = 0
while j < 3:
    print(j)
    j += 1  # 关键：更新循环变量，依次打印 0, 1, 2
```

## 二、`while` 实例：累加 0 到 10
- 目标：依次打印 `0 = 0`、`0 + 1 = 1`、……、`0 + 1 + ... + 10 = 55`
- 思路：用变量 `total` 保存累计的和，用字符串 `expression` 保存算式，每轮更新并打印

```python
# 累加 0 到 10，并打印每一步的算式
i = 0
total = 0
expression = ""  # 保存算式的字符串，初始为空

while i < 11:
    if i == 0:
        expression = "0"            # 第 0 轮：算式就是 "0"
    else:
        expression += " + " + str(i)  # 之后每轮追加 " + i"
    total += i                      # 把 i 累加到 total
    print(expression + " = " + str(total))
    i += 1                          # 更新循环变量，避免死循环
```

## 三、`while` 实例：九九乘法表（嵌套循环）
- 嵌套循环：循环里面再写一个循环
- 外层 `i` 控制行（1~9），内层 `j` 控制列（1~9）
- 规律：外层每走 1 步，内层要完整跑完 1 到 9

```python
# 九九乘法表：嵌套 while 循环
i = 1
while i <= 9:        # 外层循环：行 i 从 1 到 9
    j = 1
    while j <= 9:    # 内层循环：列 j 从 1 到 9
        print(f"{i} x {j} = {i*j}")
        j += 1
    i += 1           # 内层完整跑完一轮后，外层才加 1
```

## 四、列表 List
- 列表：一组排好“队伍”的数据，用方括号 `[]`，例如 `[90, 85, 77]`
- 特点：有顺序、可以重复、可以修改
- 索引 Indexing：每个元素有编号，**从 0 开始**，用 `列表[编号]` 取出元素
- `range()`：快速生成一串连续整数，常配合 `list()` 转成列表
  - `range(5)` → 0, 1, 2, 3, 4（从 0 开始，到 5 为止但**不包含 5**）
  - `range(1, 6)` → 1, 2, 3, 4, 5（从 1 开始，到 6 为止但不包含 6）

```python
scores = [90, 85, 77, 60]  # 列表：方括号 []，元素用逗号分隔
print(scores)              # 打印整个列表
print(scores[0])           # 索引从 0 开始：取出第 1 个元素 90
print(scores[2])           # 取出第 3 个元素 77

scores[3] = 88             # 修改第 4 个元素
print(scores)

print(len(scores))         # len()：列表里有几个元素，输出 4

print(list(range(5)))      # range(5)：0~4（从 0 开始，不包含 5）
print(list(range(1, 6)))   # range(1, 6)：1~5（不包含 6）
```

## 五、元组 Tuple
- 元组：和列表类似的一串数据，用圆括号 `()`，例如 `("Jim", 20)`
- 特点：有顺序、**不能修改**（创建后就固定了）
- **可变与不可变（mutable / immutable）**：
  - mutable（可变的）：创建后还能修改内容，例如**列表**（可以改元素、加元素）
  - immutable（不可变的）：创建后内容固定、不能修改，例如**元组**、字符串、数字
- 索引用法和列表一样：`元组[编号]`

```python
student = ("Jim", 20)  # 元组：圆括号 ()
print(student)
print(student[0])      # 取出第 1 个元素 "Jim"
print(student[1])      # 取出第 2 个元素 20

# student[0] = "Tom"   # 会报 TypeError！元组创建后不能修改
```

## 六、集合 Set
- 集合：一组**不重复**的数据，用花括号 `{}`，例如 `{1, 2, 3}`
- 特点：没有顺序、不重复（重复的元素会自动去掉）

```python
numbers = {3, 1, 2, 3, 1}  # 集合：花括号 {}
print(numbers)             # 重复的 3 和 1 只保留一个

numbers.add(4)             # 用 add() 添加元素
print(numbers)
```

## 七、字典 Dictionary
- 字典：一组“键值对”(key: value)，像查词典一样，用花括号 `{}`，例如 `{"name": "Jim", "age": 20}`
- 特点：通过“键”(key) 查找对应的“值”(value)
- 用 `字典[键]` 取出值，也可以修改值

```python
student = {"name": "Jim", "age": 20}  # 字典：键: 值
print(student)
print(student["name"])   # 通过键取出值："Jim"
print(student["age"])    # 20

student["age"] = 21      # 通过键修改值
print(student)
```

## 八、`for` 循环
- `for` 循环：对一个序列（如 `range()`、列表）里的**每个元素**依次执行一次
- 与 `while` 的相似之处：都是重复执行一段代码，也都可以用 `break` / `continue`
- 与 `while` 的区别：
  - `for`：循环次数**一开始就确定**（序列里有几个元素就循环几次），不需要自己写 `i += 1`
  - `while`：循环次数**由条件决定**，要自己更新条件，写不对会变成死循环

```python
# 同一个任务：打印 0 到 4，两种写法对比

# while 写法：要自己初始化条件、自己更新条件
i = 0
while i < 5:
    print(i)
    i += 1

# for 写法：不需要 i += 1，range(5) 自动给出 0~4
for i in range(5):
    print(i)

# for 也可以直接遍历列表里的每个元素
for score in [90, 85, 77, 60]:
    print(score)
```

## 今日小结

- `while` 实例：累加 0~10；九九乘法表（嵌套循环：外层走一步，内层跑完一整轮）
- 四种数据结构：
  - 列表 `[]`：有序、可修改
  - 元组 `()`：有序、不可修改
  - 集合 `{}`：无序、不重复
  - 字典 `{键: 值}`：按键查值
- 索引：编号从 0 开始，用 `[编号]` 取出元素
- mutable（可变）：列表；immutable（不可变）：元组、字符串、数字
- `for` 循环：遍历序列每个元素，次数确定、不用自己写 `i += 1`；`while` 由条件控制
