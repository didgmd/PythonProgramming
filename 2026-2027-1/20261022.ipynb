## 一、函数 Function
- 函数：把一段完成特定任务的代码“打包”起来，起一个名字，需要时直接使用
- 函数需要先**定义（Define）**、后**调用（Call）**：
  - 定义：`def 函数名(参数):`，下面缩进的代码是函数体（注意冒号和缩进）
  - 调用：写 `函数名(实际值)`，例如 `add5(5)`
- `return`：把计算结果“交还”给调用的地方；没有 `return` 的函数只是做一件事（如打印）
- 好处：写一次、用多次，避免重复代码

### 例1：有 `return` 的函数（返回计算结果）

```python
# 定义函数：输入一个数，返回它加 5 的结果
def add5(x):          # x 是参数（parameter），代表调用时传入的值
    return x + 5      # return 把结果交还给调用处

def mul2(num):        # 参数名叫 num，与 add5 的 x 互不影响
    return num * 2

print(add5(5))        # 调用：传入 5，返回 10 并打印
print(mul2(5))        # 传入 5，返回 10 并打印
```

### 例1b：函数的嵌套调用（函数套函数）
- 一个函数的结果可以直接传给另一个函数，例如 `mul2(add5(10))`
- 执行顺序：**先算里面、再算外面**（先 `add5(10)` 得 15，再 `mul2(15)` 得 30）

```python
# 复用例1的两个函数，演示嵌套调用
def add5(x):
    return x + 5

def mul2(num):
    return num * 2

print(mul2(add5(10)))  # 先算里面 add5(10)=15，再算外面 mul2(15)=30
print(add5(mul2(10)))  # 先算 mul2(10)=20，再算 add5(20)=25
```

### 例2：没有 `return` 的函数（直接做一件事：打印）

```python
# 没有 return 的函数：直接打印一句话
def print_hello(name):
    print(f"Hello {name}!")  # 参数 name 嵌入到 f-string 中

print_hello("Tom")    # 输出 Hello Tom!
print_hello("Lucy")   # 输出 Hello Lucy!
```

### 例3：参数是列表的函数（配合 `sum()` 和 `len()`）

```python
# 参数也可以是列表：计算列表中所有数的平均值
list1 = [123, 234, 345, 456, 567]
list2 = [678, 789, 890, 901]

def take_average(numbers: list):      # 参数 numbers 是一个列表
    return sum(numbers) / len(numbers)  # sum() 求和，len() 数个数，相除得平均值

print(take_average(list1))   # 输出 345.0
print(take_average(list2))   # 输出 814.5
```

## 二、函数参数的默认值（Default value）
- 定义函数时可以给参数一个**默认值**：`def 函数名(参数=默认值):`
- 调用时**不传**这个参数，就自动使用默认值；**传了**就用传入的值
- 可以用 `参数名=值` 的方式指定传哪个参数（关键字参数），顺序不重要
- 注意：有默认值的参数要放在没有默认值的参数**后面**

```python
# 计算折后含税价：discount（折扣%）和 tax（税%）都有默认值
def calculate_price(price, discount=0, tax=20):
    before = price * (1 - discount / 100)   # 先算折扣价
    return before * (1 + tax / 100)         # 再加税

print(calculate_price(30))                      # 都用默认值：30 加 20% 税 → 36.0
print(calculate_price(30, tax=10))              # 指定 tax=10 → 33.0
print(calculate_price(30, discount=10, tax=10)) # 打 9 折再加 10% 税 → 29.7
```

## 今日小结
- 函数：先定义（`def 函数名(参数):`）、后调用（`函数名(值)`），注意冒号和缩进
- `return` 把结果交还给调用处；没有 `return` 的函数只是做一件事（如打印）
- 参数可以是数字、字符串，也可以是列表
- 函数写一次、用多次，避免重复代码
- 嵌套调用：`mul2(add5(10))`，先算里面、再算外面
- 默认值参数：不传就用默认值；可用 `参数名=值` 指定传参（关键字参数）
