#### type() 函数
1. 获取圆括号内 量 的类型
2. 结果会以 `<class 'type'>` 形式来输出

```python
print(type(10))  # <class 'int'>
print(type("Hello"))  # <class 'str'>
print(type(int("10")))  # <class 'int'>
print(type(str(int("10"))))  # <class 'str'>
```
