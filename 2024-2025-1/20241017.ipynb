#### 对字符串进行索引(Indexing)
1. `s = "Hello"`, `s[0]`为`H`, `s[3]`为`l`
2. 索引从0开始
3. -1对应最后一个字符 `s[-1]`

```python
s = "Hello World"
print(s[0])
print(s[3])
print(s[-1])
# print(s[-3])  # -3对应的字符是?
```

#### 字符串的截取(Slicing)
1. `s2 = "Shenyang"`, `s2[1:3]`为`he`
2. 冒号左侧是开始点，右侧是结束点+1
3. `s2[:n]`表示从最左侧到第n-1个,等同于`s2[0:n]`
4. `s2[n:]`表示从左侧第n个到最右侧,等同于`s2[n:len(s2)]`

```python
s2 = "Shenyang"
print(s2[1:3])
print(s2[:3])
print(s2[3:])
print(s2[3:len(s2)])
# print(s2[3:-1])  # 结果是?
```

#### input()函数
1. 通过`input()`函数获取的输入永远是字符串类型
2. 如果想输入数字,则需要进行类型转换`int(input())`

```python
a = input()  # 输入10
print(type(a))  # <class 'str'>
b = int(a)
print(type(b))  # <class 'int'>
```

```python
name = input("Please enter your name: ")
age = int(input("Please enter your age: "))  # 由于input()函数获取的输入为字符串,因此需要将其转换为数字才能进行后面的运算
print(f"Hello {name}, you were born in either {2024 - age} or {2024 - age + 1}.")
```
