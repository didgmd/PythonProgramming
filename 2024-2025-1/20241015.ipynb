#### 数据类型 Data Type
1. 整型 `int`
2. 浮点型 `float` (小数)
3. 字符串 `str` (string)
4. 布尔型 `bool`: `True` `False`

```python
print(type(4.5))  # <class 'float'>
```

```python
print(type("hello"))  # <class 'str'>
```

```python
print(type(7))  # <class 'int'>
```

```python
print(type(7+4.5))  # <class 'float'>
```

```python
print(type(True))  # <class 'bool'>
```

```python
print(1+True)  # 2
```

```python
print(True+False)
```

#### 算数运算符 
1. 四则运算: `+` `-` `*` `/`
2. 求余: `%`
3. 幂: `**`

```python
print(5 % 3)
```

```python
print(2 ** 8)
```

#### 关系运算符
1. `>` `<` `>=` `<=`
2. `==` `!=`

```python
print(4<11)  # 关系表达式 -> 布尔
```

```python
print(3==3)  # 与 = 区分
```

```python
# 优先级：算数运算符 高于 关系运算符
print(10+5 == 3*5)  # True
```

```python
name = input("Please enter your name: ")
age = input("Please enter your age: ")
print(f"Hello, your name is {name}, your age is {age}")
print(type(name))  # <class 'str'>
print(type(age))  # <class 'str'>
```

#### 强制类型转换
1. 通过`input()`函数获取的输入,永远是`str`类型,即字符串
2. 对于字符串类型的数字,可以通过`int()`函数强制转换为整型

```python
print(type(age))
print(type(int(age)))
```
