#### 列表(List)的基本操作
1. 创建列表 `list = ['a','b','c','d','e']`
2. 添加元素 `list.append()`
3. 删除元素 `list.remove('b')` `del list[1]` `list.pop()` `pop`返回删除的元素，因此可以用来赋值 `a = list.pop()`
4. 修改元素 `list[0] = 1`
5. 查找元素 `list.index('b')`
6. 插入元素 `list.insert(2,'f')`
7. 排序 `list.sort()` `list.sort(reverse=True)`
8. 列表反转 `list.reverse()`

```python
# 查找列表元素
list_1 = ['a', 'b', 'c', 'd', 'e']
print(list_1.index('c'))
```

```python
# 插入元素
list_2 = ['a', 'b', 'c', 'd', 'e']
print(list_2)
list_2.insert(2, 'f')
print(list_2)
```

```python
# 删除元素
list_3 = ['a', 'b', 'c', 'd', 'e', 'f']
print('initial\t\t\t',list_3)
list_3.remove('c')
print('after remove \'c\'',list_3)
del list_3[1]
print('after del 1\t\t',list_3)
list_3.pop()
print('after pop()\t\t',list_3)
list_3.pop(1)
print('after pop(1)\t',list_3)
```

```python
# 二重列表(2 dimensional list)
list_4 = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print(list_4)
print(list_4[1])
print(list_4[1][1])
```

#### Immutable and Mutable(不可变和可变)
1. 不可变对象: `tuple`(元组)
2. 可变对象: `list`(列表) `dict`(字典) `set`(集合)

#### 元组(Tuple)
1. 创建元组 `tuple = ()`
2. 修改/引用元组元素 `tuple[0]`
3. 删除元组 `del tuple`
4. 元组的不可变性：元组的元素不可添加/移除，但元组本身可以删除
5. Unpacking(解包) `(a, b, c) = tuple`

```python
# 创建元组
tuple_1 = ('a', 'b', 'c', 'd', 'e')
print(tuple_1)

# 修改/引用元组元素
print(tuple_1[0])

# 解包
(A, B, C, D, E) = tuple_1
print(A, B, C, D, E)

# 删除元组
del tuple_1
```

#### 集合(Set)
1. 创建集合 `set = {}` 或 `set = set()`
2. 添加元素 `set.add()`
3. 删除元素 `set.remove()`
4. 集合的元素不可重复
5. 集合是无序的
5. `set.union()` 并集
6. `set.intersection()` 交集
7. `set.difference()` 差集

```python
# 创建集合
set_1 = {'a', 'b', 'c', 'd', 'e'}
print(set_1)

# 添加元素
set_1.add('f')
print(set_1)

# 删除元素
set_1.remove('f')
print(set_1)

# 并集
set_2 = {'a', 'b', 'c', 'd', 'e'}
set_3 = {'c', 'd', 'e', 'f', 'g'}
print(set_2.union(set_3))

# 交集
print(set_2.intersection(set_3))

# 差集
print(set_2.difference(set_3))
```

#### 字典(Dictionary)
1. 创建字典 `dict = {}` 或 `dict = dict()`
2. 添加元素 `dict['key'] = value`
3. 删除元素 `del dict['key']`
4. 修改元素 `dict['key'] = new_value`
5. 查找元素 `dict['key']`
6. 字典的键不可重复

```python
# 创建字典
dict_1 = {
    'a': 1, 
    'b': 2, 
    'c': 3, 
    'd': 4, 
    'e': 5}
print(dict_1)

# 添加元素
dict_1['f'] = 6
print(dict_1)

# 删除元素
del dict_1['f']
print(dict_1)

# 修改元素
dict_1['a'] = 0
print(dict_1)

# 查找元素
print(dict_1['a'])
```
