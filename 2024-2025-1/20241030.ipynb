#### 递归(Recursion)
- 递归是一种解决问题的方法，它是将问题分解为更小的子问题，直到问题的规模小到可以被很简单的解决。
- 递归一定要有一个终止条件，否则会陷入无限循环。

```python
# 阶乘(Factorial)
def fact(n):
    if n == 1:
        return 1
    return n * fact(n - 1)

print(fact(5))
```

```python
# 斐波那契数列(Fibonacci)
def fib(n):
    if n == 1 or n == 2:
        return 1
    return fib(n - 1) + fib(n - 2)

for i in range(1, 11):
    print(fib(i), end=' ')
```
