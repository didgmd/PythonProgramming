#### 模块(Module)
- 在Python中,文件即模块
- 使用`import`可以导入模块
- 使用`.`可以调用在该模块中定义的函数
- 使用`as`可以给导入的模块或函数重命名
- 使用`from`可以从模块中导入特定函数

```python
import math
print(math.pi)
print(math.sqrt(16))
print(math.ceil(3.3))  # 向上取整
print(math.floor(3.8))  # 向下取整
```

```python
import datetime
print(datetime.datetime.now())
```

```python
import random
print(random.randint(3, 9))
mylist = ['apple', 'banana', 'cherry']
print(random.choice(mylist))
```

```python
# 从模块中导入特定函数
from random import randint,choice
print(randint(3, 7))
print(choice(['a', 'b', 'c']))
```

```python
# 在 cmd(命令提示符) 中输入以下命令安装模块
# python -m pip install --upgrade pip -i https://pypi.tuna.tsinghua.edu.cn/simple
# python -m pip install numpy -i https://pypi.tuna.tsinghua.edu.cn/simple
# python -m pip install matplotlib -i https://pypi.tuna.tsinghua.edu.cn/simple
import matplotlib.pyplot as plt  # 导入matplotlib.pyplot模块并重命名为plt
import numpy as np  # 导入numpy模块并重命名为np
x = np.arange(1, 100, 0.1)  # 生成1到100的等差数列,步长为0.1
# print(x)
y = np.sin(x)  # 计算x的正弦值
# print(y)
plt.plot(x, y)  # 绘制图形
plt.show()  # 显示图形
```

```python
# 以写入(write)模式打开文件(如果文件不存在则创建文件),其他常用模式还有'r'(读取)和'a'(追加)
file = open('example.txt', mode='w')  

# 写入文件
file.write('Hello, World!\n')
file.write('This is our new text file\n')

# 关闭文件
file.close()
```

```python
# 以读取(read)模式打开文件(文件必须存在)
file = open('example.txt', mode='r')  
symbols = file.read()  # 按字符读取文件内容
for symbol in symbols:
    print(symbol)

# 关闭文件
file.close()
```

```python
# 以读取(read)模式打开文件(文件必须存在)
file = open('example.txt', mode='r')  # 以读取(read)模式打开文件
lines = file.readlines()  # 按行读取文件内容
for line in lines:
    print(line)

# 关闭文件
file.close()
```
