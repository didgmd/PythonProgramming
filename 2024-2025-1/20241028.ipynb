#### 内置函数
- `max(<class 'list'>)` 返回列表中的最大值
- `min(<class 'list'>)` 返回列表中的最小值

```python
list1 = [1, 2, 3, 4, 5]
list1_max = max(list1)
print(list1_max)
list1_min = min(list1)
print(list1_min)
```

#### 内置函数
- `<class 'list'>.sort()` 对列表进行从小到大排序

```python
list2 = [5, 4, 3, 2, 1]
list2.sort()  # 默认从小到大排序
print(list2)
list2.sort(reverse=True)  # 从大到小排序
print(list2)
```

#### 函数(Function)
- 函数是一段可以重复调用的代码块，可以接受参数，也可以返回值。
- 函数的定义以`def`关键字开始，后面是函数名和圆括号`()`，圆括号中可以包含参数名，多个参数之间用逗号分隔。
- 函数体是缩进的代码块，函数体中可以包含任意数量的语句。
- 函数的定义(statement)通常放在**程序的开头**，这样可以在程序的任何地方调用(call)函数。
- 先定义后调用。

```python
# 定义函数
def add5(x):  # x是形参(parameter)
    y = x + 5  # 函数体
    return y  # 返回值

result1 = add5(10)  # 10是实参(argument)
print(result1)
result2 = add5(15)  # 15是实参(argument)
print(result2)
```

#### 形参(Parameter)和实参(Argument)
- 形参是函数定义中的参数，实参是函数调用时传递给函数的参数。
- 实参可以是常量、变量、表达式、函数等。
- 实参传递给形参时，实参的值会被赋值给形参，形参的值在函数体中可以使用。

```python
def maximum(x, y):
    if x > y:
        return x
    else:
        return y

result3 = maximum(10, 15)
print(result3)
```

```python
def add(x, y):
    z = x + y
    return z

result4 = add(10, 15)
print(result4)
```

```python
# 当函数没有返回值时，返回None
def print_hello():
    print('Hello, World!')

result5 = print_hello()  # 函数没有返回值
print(result5)
```

#### Scope(作用域)
- 变量的作用域(scope)是指变量的有效范围。
- 全局变量(global variable)是在函数外部定义的变量，可以在函数内部使用。
- 局部变量(local variable)是在函数内部定义的变量，只能在函数内部使用。
- 函数内部的变量不会影响函数外部的变量。

```python
x = 10  # 全局变量
def print_x():
    print(x)  # 函数内部使用全局变量

print_x()
```

```python
def print_y():
    y = 20  # 局部变量
    print(y)  # 函数内部使用局部变量

print_y()
print(y)  # NameError 函数外部不能使用局部变量
```
