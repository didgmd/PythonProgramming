```python
# 以5cm为间隔,分别统计班级男,女同学的身高分布情况.例如:身高175-180cm的男生有b名,身高170-175cm的女生有g名.
B_below_170, B_171_175, B_176_180, B_above_180 = 0, 0, 0, 0
G_below_160, G_161_165, G_166_170, G_above_170 = 0, 0, 0, 0

while True:
    to_continue = input("Enter y for continue, n for exit")
    if to_continue == 'y':
        print("continue counting\n")
    elif to_continue == 'n':
        print("Exit counting, printing the result")
        break
    else:
        print("Invalid input")
        continue
    
    gender = input("Please enter the gender, B for boys, G for girls")
    height = float(input("Please enter the height"))
    
    if gender == 'B':
        if height < 170:
            B_below_170 += 1
        elif 171 <= height <= 175:
            B_171_175 += 1
        elif 176 <= height <= 180:
            B_176_180 += 1
        elif height > 180:
            B_above_180 += 1
    elif gender == 'G':
        if height < 160:
            G_below_160 += 1
        elif 161 <= height <= 165:
            G_161_165 += 1
        elif 166 <= height <= 170:
            G_166_170 += 1
        elif height > 180:
            G_above_170 += 1
    else:
        print("Invalid input")
        continue

print(f"Boys "
      f"below 170: {B_below_170}, "
      f"between 171 and 175: {B_171_175}, "
      f"between 176 and 180: {B_176_180}, "
      f"above 180: {B_above_180}")

print(f"Girls "
      f"below 160: {G_below_160}, "
      f"between 161 and 165: {G_161_165}, "
      f"between 166 and 160: {G_166_170}, "
      f"above 170: {G_above_170}")
```
