```python
# 输出1到100之间的素数(prime number).
for num in range(1, 101):
    is_prime = True
    if num < 2:
        continue
    
    for i in range(2, int(num/2) + 1):
        if num % i == 0:
            is_prime = False
            break
    
    if is_prime:
        print(num,"is a prime number.")
```

#### 死循环(Infinite Loop)
- 无限循环,程序不会停止
- 通常是由于逻辑错误导致的,需要避免死循环

```python
# 死循环
while True:
    print('Hello, infinite loop!')

# 死循环
a = 1
while a <= 10:
    print(a)
    a -= 1  # i = i - 1
```

#### 列表(List)
- 有序的元素集合,用`[ ]`表示,元素之间用逗号分隔
- 列表中的元素可以是不同类型的数据,可以重复
- 列表中的元素可以通过索引访问
- 列表是可变的,可以修改,添加,删除元素
- 列表可以用`for`循环遍历

```python
list_1 = [1, 23, 'hello', 3.14, True]
print(list_1)
print(list_1[2])
print(list_1[-1])
print(list_1[1:3])  # 注意:切片操作(Slicing)后的结果是一个新的列表
```

```python
for item in list_1:
    print(item)
```

#### 内置函数
- `sys.getsizeof(object)`:返回对象的字节(Byte)大小,需要`import sys`

```python
import sys
print(sys.getsizeof(3.14))  # float类型的对象占用24个字节(24 Bytes)
print(sys.getsizeof(10))  # int类型的对象占用28个字节(28 Bytes)
print(sys.getsizeof(True))  # bool类型的对象占用28个字节(28 Bytes)
print(sys.getsizeof('hello'))  # str类型的对象占用46个字节(46 Bytes)
print(sys.getsizeof(range(1, 5)))  # range类型的对象占用48个字节(48 Bytes)
print(sys.getsizeof(list(range(1, 5))))  # list类型的对象占用96个字节(96 Bytes)
```

```python
my_range = range(10, 1, -1)
print(my_range.start)  # 范围的起始值
print(my_range.stop)  # 范围的结束值
print(my_range.step)  # 范围的步长
print(my_range.index(7))  # 返回指定值的索引
```

```python
# 用`for`遍历字符串,输出所有数字
str_1 = 'hello numbers 767826 numbers 349714 goodbye numbers'
for x in str_1:
    if '0' <= x <= '9':
        print(x)
```

```python
# 用`for`遍历数字字符串,找到最大的2位数字
str_2 = "12576287136512634182759"
max_2_digit = None
for i in range(len(str_2) - 1):
    current_num = int(str_2[i:i+2])
    if max_2_digit:
        if current_num >= max_2_digit:
            max_2_digit = current_num
    else:
        max_2_digit = current_num

print(f"The maximum 2 digit number is {max_2_digit}")
```

#### 处理例外情况(Exception)
- `try ... except ...`
- 尝试`try`后的语句,无法执行时,运行`except`后的语句

```python
str_3 = "hello"
try:
    num = float(str_3)
except:
    print(f"Cannot convert {str_3} to float type")
```
