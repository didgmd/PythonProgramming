```python
# 练习: 如何修改程序，让其能够显示 1st, 2nd, 3rd, 4th, 11st, ... ...
s = input("Please enter a string: ")
i = 0
while i < len(s):
    digit = int(str(i+1)[-1])
    
    if digit == 1:
        print(f"The {i+1}st character of {s} is {s[i]}")
    elif digit == 2:
        print(f"The {i+1}nd character of {s} is {s[i]}")
    elif digit == 3:
        print(f"The {i+1}rd character of {s} is {s[i]}")
    else:
        print(f"The {i+1}th character of {s} is {s[i]}")
    
    i += 1  # i = i + 1
```

#### 嵌套 if 语句(Nested if statement)
- `if condition: if condition: ... else: ... else: ...`
- `if condition: ... else: if condition: ... else: ...`


```python
year = int(input("Please enter a year: "))

if year % 4 == 0 and year % 100 != 0:
    print(f"{year} is a leap year")
else:
    if year % 400 == 0:
        print(f"{year} is a leap year")
    else:
        print(f"{year} is not a leap year")
```
