#### 逻辑运算符(Logical operator)
1. `and`: 只有`True and True`时,结果为`True`,其余为`False`
2. `or`: 只有`False or False`时,结果为`False`,其余为`True`
3. `not`: `not True`结果为`False`,反之为`True`

```python
x = 7
print(not(x < 5 and x < 10))
```

```python
print(3 < 5 and 6.2 != 5)
```

#### 条件分支(Condition branching)
1. `if condition: ... else: ...`
2. `if condition: ... elif condition: ... elif condition: ... else ...`

$ax^2 + bx + c = 0$

```python
import math
a, b, c = 3, 8, 5
disc = b**2 - 4 * a * c
if disc < 0:
    print("没有实根")
else:
    print(f"存在实根: {(-b + math.sqrt(disc)) / (2 * a)}, {(-b - math.sqrt(disc)) / (2 * a)}")       
```

```python
x = 3
y = 5
if x > y:
    print("x is larger than y")
elif x < y:
    print("x is smaller than y")
else:
    print("x is equal to y")
```

```python
grade = 85
if grade < 60:
    print("E")
elif grade >= 60 and grade < 70:
    print("D")
elif grade >= 70 and grade < 80:
    print("C")
elif grade >= 80 and grade < 90:
    print("B")
elif grade >= 90 and grade <= 100:
    print("A")
elif grade > 100:
    print("Error")
```

#### 循环(Loop)
1. `while condition: ...`

```python
s = input("Please enter a string: ")
i = 0
while i < len(s):
    print(f"The {i+1} character of {s} is {s[i]}")
    i += 1  # i = i + 1

# 练习: 如何修改程序，让其能够显示 1st, 2nd, 3rd, 4th, 11st, ... ...
```
