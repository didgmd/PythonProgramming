```python
# 以追加(append)模式打开文件(如果文件不存在则创建文件)
file = open('example.txt', mode='a')  # 以追加(append)模式打开文件
file.write('This is our new text file\n')  # 写入文件

# 关闭文件
file.close()

# 以读取(read)模式打开文件(文件必须存在)
file = open('example.txt', mode='r')  # 以读取(read)模式打开文件
lines = file.readlines()  # 按行读取文件内容
for i,line in enumerate(lines):
    print(i, end=' ')
    print(line)

# 关闭文件
file.close()
```

```python
from functools import reduce
def count_words(filename):
    with open(filename, "r") as f:
        lines = f.readlines()
        print(lines)
    
    # 去除每行首尾的空格和特殊符号,然后合并成一个列表
    stripped_lines = map(lambda s: s.strip(), lines)
    # print(list(stripped_lines))
    
    # 将每行分割成单词,然后将所有单词合并成一个列表
    split_lines_words = map(lambda s: s.split(), stripped_lines)
    # print(list(split_lines_words))
    
    # 将所有单词合并成一个列表
    flatten_words = reduce(lambda a, b: a + b, split_lines_words)
    print(flatten_words)
    # print(len(flatten_words))
    return sum([1 for word in flatten_words if len(word) > 0])

def main():
    print(count_words("example.txt"))

if __name__ == "__main__":
    main()
```

#### 类(Class)
- 类是一种用户自定义的数据结构,用于封装数据和方法
- 类是对象的模板,对象是类的实例
- 类由属性和方法组成
- 属性是类的数据成员,方法是类的函数成员
- 类的属性和方法通过点(.)运算符访问

```python
# 创建一个类
class Pet:
    def talk(self):
        print("Hi, I am a instance of Pet class")
    
    def speak(self):
        print("I don't know how to speak")

if __name__ == "__main__":
    pet_1 = Pet()
    pet_2 = Pet()
    print(type(pet_1))
    pet_1.talk()
    pet_2.talk()
    pet_1.speak()
    pet_2.speak()
```

`if __name__ == "__main__":`是Python的一个惯用法,表示当模块被直接运行时,代码块将被运行,当模块被导入时,代码块不被运行


```python
class Person:
    # 类的构造函数(初始化)
    def __init__(self, n, a):
        self.name = n  # 创建属性name并赋值
        self.age = a  # 创建属性age并赋值
        print(f"Person {self.name} is created")
    
    # Self是类的实例,用来访问类的属性和方法
    def say_hello(self):
        print(f"Hello, my name is {self.name}, I am {self.age} years old")

if __name__ == "__main__":
    person_1 = Person("Alice", 25)
    person_2 = Person("Bob", 30)
    person_1.say_hello()
    person_2.say_hello()
```
