#### 变量(Variable) & 赋值(Assignment)
1. `=` 将右侧的值赋给左侧的变量,如:`pi = 3.14`,`radius = 6`
2. 先计算`=`右侧的值,然后赋给左侧的变量,如:`circumference = 2 * pi * radius`
3. 此处的`pi`,`radius`,`circumference`都是变量,而`3.14`,`6`都是常量(Constant)

```python
pi = 3.14
radius = 6
circumference = 2 * pi * radius
print(circumference)
print(f"circumference is {circumference}")  # 两个 circumference 有什么不同？
```

```python
radius = 8  # 变量的值可以改变
print(f"circumference is {circumference}")  # 37.68
circumference = 2 * pi * radius
print(f"circumference is {circumference}")  # 50.24
```

#### 运算符(Operator) & 表达式(Expression)
1. 算数运算符:`+`,`-`,`*`,`/`,`%`,`**`,`//`
2. 算术表达式:`3 + 5`
3. 关系运算符:`>`,`<`,`>=`,`<=`,`==`,`!=`
4. 关系表达式:`5 > 3`结果为`True`,`3 > 5`结果为`False`
5. 逻辑运算符:`and`,`or`,`not`
6. 逻辑表达式:`True and False`

```python
print("True and False")
print(True and False)
print(5 > 3 or 3 < 5)
```

```python
f = 3.1
print(int(f))
```
