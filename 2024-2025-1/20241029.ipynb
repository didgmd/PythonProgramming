```python
# 练习题：编写一个函数，实现 `<class 'list'>.sort()` 的功能，对数字列表进行从小到大排序。
def sort_num_list(list_to_sort, reverse):
    length = len(list_to_sort)
    if reverse is False:
        for i in range(length):
            for j in range(i+1, length):
                if list_to_sort[i] > list_to_sort[j]:
                    temp = list_to_sort[i]
                    list_to_sort[i] = list_to_sort[j]
                    list_to_sort[j] = temp
    else:
        for i in range(length):
            for j in range(i+1, length):
                if list_to_sort[i] < list_to_sort[j]:
                    temp = list_to_sort[i]
                    list_to_sort[i] = list_to_sort[j]
                    list_to_sort[j] = temp
# 测试
list_1 = [2, 3, 1, 4, 5]
sort_num_list(list_1, reverse=False)
print(list_1)  # [1, 2, 3, 4, 5]
sort_num_list(list_1, reverse=True)
print(list_1)  # [5, 4, 3, 2, 1]
```

#### 函数调用时指定参数名
- `function(parameter=argument)`，这种调用方式可以不按照形参(parameter)的顺序传递实参(argument)，而是通过形参名来指定实参。

```python
def calculate_price(price, discount=0, tax=20):
    total_before_tax = price * (1 - discount/100)
    total_after_tax = total_before_tax * (1 + tax/100)
    return total_after_tax
# 测试
print(calculate_price(100))  # 120.0
print(calculate_price(100, tax=10))  # 110.0
print(calculate_price(100, discount=10))  # 108.0
print(calculate_price(100, discount=10, tax=10))  # 99.0
```
