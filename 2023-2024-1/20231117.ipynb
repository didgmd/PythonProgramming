#### 列表 List
1. 列表直接赋值时传递的是**列表对象** `list2 = list1`9
2. 复制列表内容至新列表 Hard Copy `list2 = list1.copy()`

```python
list1 = [1, 2, 3]
list2 = list1
list3 = list1.copy()
print(f"list1 is {list1}")
print(f"list2 is {list2}")
print(f"list3 is {list3}\n")
list2.append(4)
print(f"list1 is {list1}")
print(f"list2 is {list2}")
print(f"list3 is {list3}")
```

#### 递归 Recursion
1. 函数调用其自身

```python
def print_three_times(count):
    if count < 3:
        print('In function')
        print_three_times(count + 1)
        
print_three_times(0)
```

```python
# 阶乘
def fact(n):
    if n < 2:
        return 1
    return n * fact(n-1)

print(fact(4))
```

```python
# 斐波那契数列
def fib(n):
    if n < 2:
        return 1
    return fib(n-1) + fib(n-2)

for i in range(10):
    print(f"fib({i}) = {fib(i)}")
```

#### 拓扑 Tuple
1. 拓扑本身与元素均不可变
2. `tuple.count()`
2. `tuple.index()`

```python
car_tuple = ('Audi', 'BMW', 'Mercedes')
print(car_tuple[0])
print(car_tuple[0:2])
for car in car_tuple:
    print(car)
```

```python
# 拓扑解包 unpacking a tuple
(car1, car2, car3) = car_tuple
print(car1, car2, car3)
```

```python
def num_list_stats(num_list):
    len_list = len(num_list)
    sum_list = sum(num_list)
    avg_list = sum_list/len_list
    return (len_list, sum_list, avg_list)

list1 = [2, 4.5, 6, 9]
(len_len, sum_sum, avg_avg) = num_list_stats(list1)
print(f"Length is {len_len}")
print(f"Sum is {sum_sum}")
print(f"Average is {avg_avg}")
```

```python
ttt = ('a', 'b', 'c', 'a')
print(ttt.count('a'))
print(ttt.index('b'))
```

#### 集合 Set
1. 集合元素不可变
2. 集合可变
3. 添加元素 `set.add()`
4. 删除元素 `set.remove()`
5. 连接 `set.union()`
6. 交集 `set.intersection()`

```python
fruit_set = {'apple', 'b', 'banana', 'mango'}
fruit_set.add('cherry')
fruit_set.remove('mango')
print(fruit_set)
c = {'apple', 'b', 'c'}
d = fruit_set.intersection(c)
print(d)
e = {1, 2, 3}
f = {1, 5, 6}
g = e.intersection(f)
print(g)
```
