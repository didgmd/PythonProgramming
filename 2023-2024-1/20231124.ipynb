#### 类 Class
1. 使用`class 类名`创建类
2. 在类中可以创建函数，函数的第1个参数必须为`self`
3. 一个类中的`__init__`函数会在创建类对象时直接运行，其他函数需要使用类对象进行调用
4. 通过`self`可以传递参数

```python
# 创建Pet类
class Pet:
    def __init__(self, s, n):
        self.name = s
        self.age = n
        print(f'Pet object: {self.name} is created')
        
    def talk(self):
        print(f'Hi, I am an instance of pet: {self.name}, my age is {self.age}')

pet_one = Pet("Dog", 5)  # 创建第一个Pet类型的对象，创建时会自动运行__init__()函数
pet_two = Pet("Cat", 6)  # 创建第二个Pet类型的对象，创建时会自动运行__init__()函数
pet_one.talk()  # 调用Pet类下的talk()函数
pet_two.talk()  # 调用Pet类下的talk()函数
```
