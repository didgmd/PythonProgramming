#### File 读取当前位置的(文件夹)文件
1. `file=open("example.txt")` `lines=file.read()` `file.close()`
2. `with open("example.txt") as file:`

```python
file = open("calculator.py")
lines = file.read()
file.close()
print(lines)
```

```python
with open("calculator.py") as file:
    lines = file.read()
    file.close()
print(lines)
```

```python
file = open("calculator.py")
line = file.readline()
file.close()
print(line)
```

```python
file = open("calculator.py")
lines = file.readlines()
file.close()
print(lines)
```

#### 读取文件
1. 只读 read `open("file.txt", "r")`
2. 不指定方式时为只读 `open("file.txt")`
3. 可写 write `open("file.txt", "w")`

```python
file = open('hello.py', 'w')
file.write("print(\"Hello World\")")
file.close()
```

#### 转义字符
1. 新的一行 newline `\n`
2. 转义字符占1个字符长度

```python
print("Hello\nWorld")
a = "Hello\nWorld"
print(len(a))
```

```python
file = open("calculator.py")
line_count = 0
for line in file:
    line_count += 1
    if line.startswith("def"):  # 以某字符串起始的行
        # print(line)
        print(line.rstrip())
print(line_count)
file.close()
```
