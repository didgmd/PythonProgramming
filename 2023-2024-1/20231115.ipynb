#### 排序 Sorting
1. Ascending 从小到大 `list.sort()`

```python
list1 = [5, 3, 8, 1, 2]

print(max(list1))
list1.sort()
print(list1)
print(list1[-1])
```

#### 函数 Function
1. 函数定义：`def add5(x):`和返回值`return x+5`
2. 函数调用：`add5(1)`
3. 没有返回值的函数：`def print_hello(name):`和`print(f"Hello {name}")`
4. 形参 Parameter：函数定义时的参数
5. 实参 Argument：函数调用时的参数
6. 变量作用范围 Scope：全局 Global，局部 Local
7. 空函数 `pass`

```python
def add5(x):
    return x+5
print(add5(1))

def print_hello(name):
    print(f"Hello {name}")
    
print_hello("HOHO")
```
