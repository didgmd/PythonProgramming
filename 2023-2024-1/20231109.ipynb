```python
i = 0  # i的作用是计数器

while i < 10:
    print(f"Hello World! {i}")
    i += 1  # i = i + 1

print("End of while loop")
```

```python
i = 10  # i的作用是计数器

while i > 0:
    print(f"Hello World! {i}")
    i -= 1  # i = i - 1

print("End of while loop")
```

```python
num = 1

while num != 0:
    num = int(input("Please enter a number: "))
    print(f"The number you've entered is {num}")
    
print(f"End of while loop due to entered number is {num}")
```

```python
i = 1  # 第一层循环的计数器

while i < 13:
    j = 1  # 第二层循环的计数器
    while j < 12:
        print(f"{i} * {j} is {i*j}")
        j += 1
    i += 1

print("End of while loop")
```
