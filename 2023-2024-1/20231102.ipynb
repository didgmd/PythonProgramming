#### 变量赋值 Variable Assignment
1. 变量名 `=` 值

```python
pi = 3.14
# radius = 6
radius = int(input("Please enter the radius.\n"))
circumference = 2 * pi * radius
print(f"Circumference is {circumference}.")
```

#### 字符串索引 String Indexing

```python
str1 = "String Indexing"

# 正向索引
print(str1[0])
print(str1[2])
print(str1[0:2])  # 注意：结尾是数量
print(str1[:2])

# 反向索引
print(str1[-1])
```
