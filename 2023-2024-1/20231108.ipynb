```python
x = int(input("Please enter a number:"))

THRESHOLD = 10  # 字符型常量，threshold: 阈值

if x > THRESHOLD:
    print(f"{x} is larger than {THRESHOLD}")
elif x == THRESHOLD:
    print(f"{x} is equal to {THRESHOLD}")
else:
    print(f"{x} is smaller than {THRESHOLD}")
```

```python
x = int(input("Please enter a number:"))

if x % 2 == 0:
    print(f"{x} 能被2整除")
elif x % 3 == 0:
    print(f"{x} 能被3整除")
elif x % 5 == 0:
    print(f"{x} 能被5整除")
elif x % 7 == 0:
    print(f"{x} 能被7整除")
elif x % 11 == 0:
    print(f"{x} 能被11整除")
else:
    print(f"{x} 很顽固")
```

```python
x = int(input("Please enter a number:"))

# 判断输入的数字是否在1和10之间
# if (x >= 1) and (x <= 10):
#     print(f"{x} is between 1 and 10")
# else:
#     print(f"{x} is not between 1 and 10")

# 分段判断1~10, 11~20, 21~30，以及以外的数字
if (x >= 1) and (x <= 10):
    print(f"{x} is between 1 and 10")
elif (x >= 11) and (x <= 20):
    print(f"{x} is between 11 and 20")
elif (x >= 21) and (x <= 30):
    print(f"{x} is between 21 and 30")
else:
    print(f"{x} is out of range")
```

```python
x = int(input("Please enter a number: "))

if x % 3 == 0:
    if x % 5 == 0:
        print(f"{x} is divisible by 3 and 5")
    else:
        print(f"{x} is divisible by 3 but not divisible by 5")
elif x % 5 == 0:
    if x % 3 == 0:
        print(f"{x} is divisible by 5 and 3")
    else:
        print(f"{x} is divisible by 5 but not divisible by 3")
else:
    print(f"{x} is not divisible by both 3 and 5")
```
