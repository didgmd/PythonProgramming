#### 遍历列表 Traverse List
1. 直接遍历每个元素：`for x in list:`, `print(x)`
2. 通过索引遍历每个元素：`for i in range(len(list)):`, `print(list[i])`

```python
# 遍历列表 Traverse List
# 1. 直接遍历每个元素
list1 = ['a', 'b', 'c']
for x in list1:
    print(x)

# 2. 通过索引遍历每个元素
list2 = ['d', 'e', 'f']
for i in range(len(list2)):
    print(list2[i])
```

#### 列表合并与切割
1. 使用`+`可以将多个列表合并
2. 使用范围索引可以对列表进行切割 Slicing

```python
# 列表合并
a = [1, 2, 3]
b = [4, 5, 6]
c = a + b
print(c)
```

```python
# 列表切割
t = [5, 6, 7, 8, 9, 10]
print(t[0:3])  # 0为从索引号0开始，3-1为结束的索引
print(t[1:3])  # 1为从索引号1开始，3-1为结束的索引
print(t[2:3])  # 2为从索引号2开始，3-1为结束的索引
print(t[1:4])
print(t[2:5])
print(t[:5])  # :为从左侧所有元素，5-1为结束的索引，等价于print(t[0:5])
print(t[2:])  # 2为从索引号2开始，:为右侧所有元素
print(t[:])  # 所有元素，等价于print(t)
```

#### 列表操作
1. 增加元素 `list.append(a)`
2. 删除元素 `list.remove(b)`
3. 插入元素 `list.insert(index, c)`
4. 扩展列表 `list.extend(another_list)`
5. 弹出元素 `list.pop()`

```python
list3 = list()  # 定义list3为列表类型
list3.append('I')
list3.append('Love')
list3.append('SUT')
print(list3)
```

```python
list3.remove('I')
list3.remove('Love')
list3.append('Loves')
list3.append('Me')
print(list3)
```

```python
list3.insert(1, 'Always')
print(list3)
```

```python
list4 = ['Very', 'Much']
list3.extend(list4)
print(list3)
```

```python
list5 = [6, 7, 8, 9, 10]
print(list5)
list5.pop()
print(list5)
list5.pop(1)
print(list5)
```

#### 字符串分离成列表 Split
1. `string.split()` => `list`
2. 定界符 delimiter

```python
# 字符串分离成列表 Split
string1 = 'Hello World!'
list5 = string1.split()  # 以空格分离
print(list5)
```

```python
list6 = string1.split('o')  # 以字符'o'分离
print(list6)
```
