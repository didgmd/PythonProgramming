#### 常量 Constant
1. 值不发生改变 `print(123.456)`
2. 保留/关键字 `False` `class` `return` `is` `for` `if`

#### 变量 Variable
1. 变量名 `x=12.3`
2. 值可以改变 `x=100`
3. 命名规则：
3.1 由字母或下划线开始 `spam23` `_speed`
3.2 区分大小写 Case Sensitive `Spam != SPAM`

#### 赋值 Assignment
1. 赋值声明 Assignment Statement `x=2`
2. 赋值表达式 Assignment with Expression `y=x+1`
3. 先计算等号右边，然后赋值给等号左边

#### 算数表达式 Numeric Expression
1. 加减乘除
2. 指数 `**`
3. 求余 `%`
4. 整型运算 -> 取整 `8/5` => `1`

#### 运算优先级 Order of Evaluation
1. (小)括号 Parenthesis
2. 指数 Power
3. 乘除 Multiplication
4. 加减 Addition
5. Left to Right
